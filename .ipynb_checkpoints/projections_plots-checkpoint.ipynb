```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress

def load_calc(file1):
    # Load file
    calc_data = np.loadtxt(file1, skiprows=0, unpack=False)
    calc_header = ['id', 'time', 'Tair', 'snow', '0.01m', '0.08m', '0.13m', '0.24m', '0.38m', '0.53m', '0.67m', 'idk', 'idk', 'idk']
    df = pd.DataFrame(calc_data, columns=calc_header)

    # Remove the first 9 time points (spin up)
    df_filtered = df.iloc[9:].reset_index(drop=True)

    # Create a range for years 2017-2100
    start_year = 2016
    end_year = 2100
    years = list(range(start_year, end_year))

    # Assign the years as the new index
    df_filtered.index = years

    return df_filtered

def calculate_slope_and_r2(x, y):
    # Calculate linear regression
    slope, intercept, r_value, p_value, std_err = linregress(x, y)
    r2 = r_value ** 2
    return slope, r2

def find_zero_crossings(x, y):
    """Find zero crossings where the value crosses from negative to positive."""
    x = np.array(x)
    y = np.array(y)
    zero_crossings = []

    for i in range(len(y) - 1):
        if y[i] < 0 and y[i + 1] > 0:
            zero_crossing = x[i] - y[i] * (x[i + 1] - x[i]) / (y[i + 1] - y[i])
            zero_crossings.append(zero_crossing)
        elif y[i] > 0 and y[i + 1] < 0:
            zero_crossing = x[i] - y[i] * (x[i + 1] - x[i]) / (y[i + 1] - y[i])
            zero_crossings.append(zero_crossing)
    
    return zero_crossings


# Load datasets
df_ssp245_sparse = load_calc('mean_CMIP6_ssp245_deep_O_layer.txt')
df_ssp585_sparse = load_calc('mean_CMIP6_ssp585_deep_O_layer.txt')
df_ssp245_dense = load_calc('mean_CMIP6_ssp245_shallow_O_layer.txt')
df_ssp585_dense = load_calc('mean_CMIP6_ssp585_shallow_O_layer.txt')

# Inspect DataFrame columns
print("SSP2-4.5 Sparse Alder Organic Layer columns:", df_ssp245_sparse.columns.tolist())
print("SSP5-8.5 Sparse Alder Organic Layer columns:", df_ssp585_sparse.columns.tolist())
print("SSP2-4.5 Dense Alder Organic Layer columns:", df_ssp245_dense.columns.tolist())
print("SSP5-8.5 Dense Alder Organic Layer columns:", df_ssp585_dense.columns.tolist())

# Plot in a 2x2 grid
fig, axs = plt.subplots(2, 2, figsize=(14, 12), sharey=True)

# Set color cycle using shiny color palette
hex_colors = ['#1F9BF8', '#F5A623', '#7AC943', '#FF6F61', '#00BFAE', '#A57BD4']
for ax in axs.flat:
    ax.set_prop_cycle(color=hex_colors)

# Prepare to display text for slopes, R2, and zero crossings
def annotate_plot(ax, x, y, title):
    for i, depth in enumerate(y.columns):
        if depth not in y.columns:
            print(f"Column '{depth}' not found in DataFrame.")
            continue
        
        slope, r2 = calculate_slope_and_r2(x, y[depth])
        zero_crossings = find_zero_crossings(x, y[depth])
        
        if zero_crossings:
            zero_crossings_str = ', '.join(f'{int(zc)}' for zc in zero_crossings)
        else:
            zero_crossings_str = 'None'
        
        ax.annotate(f'{depth}: slope={slope:.2f}\nR2={r2:.2f}\nZero Crossing: {zero_crossings_str}', 
                    xy=(0.05, 0.95 - i * 0.1), 
                    xycoords='axes fraction', 
                    fontsize=8, 
                    verticalalignment='top')
        print(f"{title} - {depth}: Zero Crossing at {zero_crossings_str}")


# Plot each dataset with annotations
df_ssp245_sparse[['0.08m', '0.13m', '0.24m', '0.38m', '0.53m', '0.67m']].plot(ax=axs[0, 0], linewidth=2)
annotate_plot(axs[0, 0], df_ssp245_sparse.index, df_ssp245_sparse[['0.08m', '0.13m', '0.24m', '0.38m', '0.53m', '0.67m']], 'SSP2-4.5 Sparse Alder Organic Layer')
axs[0, 0].set_title('SSP2-4.5 Sparse Alder; Deep Organic Layer')

df_ssp585_sparse[['0.08m', '0.13m', '0.24m', '0.38m', '0.53m', '0.67m']].plot(ax=axs[0, 1], linewidth=2, legend=False)
annotate_plot(axs[0, 1], df_ssp585_sparse.index, df_ssp585_sparse[['0.08m', '0.13m', '0.24m', '0.38m', '0.53m', '0.67m']], 'SSP5-8.5 Sparse Alder Organic Layer')
axs[0, 1].set_title('SSP5-8.5 Sparse Alder; Deep Organic Layer')

df_ssp245_dense[['0.08m', '0.13m', '0.24m', '0.38m', '0.53m', '0.67m']].plot(ax=axs[1, 0], linewidth=2, legend=False)
annotate_plot(axs[1, 0], df_ssp245_dense.index, df_ssp245_dense[['0.08m', '0.13m', '0.24m', '0.38m', '0.53m', '0.67m']], 'SSP2-4.5 Dense Alder Organic Layer')
axs[1, 0].set_title('SSP2-4.5 Dense Alder; Shallow Organic Layer')

df_ssp585_dense[['0.08m', '0.13m', '0.24m', '0.38m', '0.53m', '0.67m']].plot(ax=axs[1, 1], linewidth=2, legend=False)
annotate_plot(axs[1, 1], df_ssp585_dense.index, df_ssp585_dense[['0.08m', '0.13m', '0.24m', '0.38m', '0.53m', '0.67m']], 'SSP5-8.5 Dense Alder Organic Layer')
axs[1, 1].set_title('SSP5-8.5 Dense Alder; Shallow Organic Layer')

# Add dotted horizontal line at y=0 for all plots
for ax in axs.flat:
    ax.axhline(0, color='black', linestyle='--', linewidth=1, alpha=0.5, zorder=0)
    ax.grid(True)

# Set common labels
axs[1, 0].set_ylabel('Ground Temperature (°C)')
axs[1, 0].set_xlabel('Year')
axs[1, 1].set_xlabel('Year')

# Adjust layout and save the figure
plt.tight_layout()
plt.savefig('ground_temp_projections_alder_O_layer_simulation_with_grids_and_stats.png', dpi=500)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def logarithmic_model(x, a, b):
    """Logarithmic model used for curve fitting."""
    return a * np.log(x) + b

def combined_model(x, log_params, slope, intercept, zero_crossing):
    """Combined log-linear model for fitting."""
    return np.where(x <= zero_crossing, 
                    logarithmic_model(x, *log_params), 
                    slope * x + intercept)

def find_zero_crossing(x, y):
    """Find the first crossing from negative to positive."""
    x = np.array(x)
    y = np.array(y)
    for i in range(len(y) - 1):
        if y[i] < 0 and y[i + 1] > 0:
            zero_crossing = x[i] - y[i] * (x[i + 1] - x[i]) / (y[i + 1] - y[i])
            return zero_crossing
    return None

def fit_log_then_linear(x, y):
    """Fit logarithmic model until zero-crossing, then a linear model."""
    zero_crossing = find_zero_crossing(x, y)

    if zero_crossing is None:
        print("No zero crossing found")
        return None, None, None, None

    # Fit logarithmic model up to zero-crossing
    x_log = x[x <= zero_crossing]
    y_log = y[x <= zero_crossing]
    try:
        log_params, _ = curve_fit(logarithmic_model, x_log, y_log)
    except:
        print("Error fitting log model")
        return None, None, None, None

    # Fit linear model after zero-crossing
    x_lin = x[x > zero_crossing]
    y_lin = y[x > zero_crossing]
    if len(x_lin) == 0:  # Check if there's any data to fit
        print("No data available for linear fit")
        return log_params, None, None, zero_crossing
    slope, intercept = np.polyfit(x_lin, y_lin, 1)

    return log_params, slope, intercept, zero_crossing

def process_and_plot(df, ax, title):
    """Process data and plot log-linear fit for each depth column."""
    x = df['time']
    
    for depth in ['0.08m', '0.13m', '0.24m', '0.38m', '0.53m', '0.67m']:
        if depth not in df.columns:
            continue  # Skip if the depth column does not exist

        y = df[depth]  # Each depth column

        # Fit log then linear model
        log_params, slope, intercept, zero_crossing = fit_log_then_linear(x, y)

        if log_params is not None:
            # Plot the result for each depth line
            ax.plot(x, y, label=f'Depth {depth}')  # Plot the actual data
            if slope is not None and intercept is not None:
                ax.plot(x, combined_model(x, log_params, slope, intercept, zero_crossing), linestyle='--')  # Fit plot

    ax.set_title(title)

def plot_all():
    fig, axs = plt.subplots(2, 2, figsize=(15, 10))
    hex_colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']

    for ax in axs.flat:
        ax.set_prop_cycle(color=hex_colors)
    
    # Apply the plotting function to each dataset
    process_and_plot(df_ssp245_sparse, axs[0, 0], 'SSP2-4.5 Sparse Alder Organic Layer')
    process_and_plot(df_ssp585_sparse, axs[0, 1], 'SSP5-8.5 Sparse Alder Organic Layer')
    process_and_plot(df_ssp245_dense, axs[1, 0], 'SSP2-4.5 Dense Alder Organic Layer')
    process_and_plot(df_ssp585_dense, axs[1, 1], 'SSP5-8.5 Dense Alder Organic Layer')

    # Set common axis labels
    axs[1, 0].set_ylabel('Ground Temperature (°C)')
    axs[1, 0].set_xlabel('Year')
    axs[1, 1].set_xlabel('Year')

    # Create a combined legend outside the plot
    handles, labels = axs[0, 0].get_legend_handles_labels()  # Get handles and labels from one of the axes
    fig.legend(handles, labels, loc='upper center', ncol=3, bbox_to_anchor=(0.5, -0.05))  # Place the legend below

    plt.tight_layout()
    plt.subplots_adjust(top=0.9, bottom=0.2)  # Adjust layout to make room for the legend
    plt.show()

# Example usage with your datasets
# df_ssp245_sparse, df_ssp585_sparse, df_ssp245_dense, df_ssp585_dense should be defined as your datasets
plot_all()
```
